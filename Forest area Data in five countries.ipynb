```python
# Forest area status in five countries
import wbdata
import matplotlib.pyplot as plt
from google.colab import files

new_countries = ["DEU", "JPN", "BRA", "ZAF", "CAN"]
new_indicators= {"AG.LND.FRST.K2":"forest_area"}

# Load the data and indicators from 2000 to 2025
n_df = wbdata.get_dataframe(
    new_indicators,
    country = new_countries,
    date = ("2000","2025")
)

# Process the data frame
n_df = n_df.reset_index()
n_df = n_df.rename(columns = {"date":"year"})
n_df["year"] = n_df["year"].astype(int)

country_info = wbdata.get_countries(new_countries)
country_names = {c['id']: c['name'] for c in country_info}

plt.figure(figsize =(12,7))

markers = ['o','s','^','D','X']
marker_idx = 0

# Iterate over unique country names present in Dataframe
for country_name in n_df["country"].unique():
  sub_df = n_df[n_df["country"] == country_name]
  plt.plot(sub_df["year"], sub_df["forest_area"],
           label = country_name,
           marker = markers[marker_idx],
           markersize = 6)
  marker_idx += 1

plt.xlabel("Year")
plt.ylabel("forest_area")

plt.title("Forest area in Country, 2000-2025", fontsize = 14)
plt.grid(True, alpha = 0.3)

plt.legend(bbox_to_anchor=(1.02, 1), loc="upper left", borderaxespad=0.)
plt.tight_layout()

file_name = "Forest area status.png"
plt.savefig(file_name, dpi=300, bbox_inches="tight")

plt.show()

files.download(file_name)

```
