```python
# ISS location temperature

import requests
iss_url = "http://api.open-notify.org/iss-now.json"
iss_response = requests.get(iss_url)
iss_data = iss_response.json()

# Extract the lat and long coordinates
iss_la = iss_data["iss_position"]["latitude"]
iss_long = iss_data["iss_position"]["longitude"]
print("received location: ", iss_la,", ",iss_long)

weather_url = "https://api.open-meteo.com/v1/forecast"
weather_params = {
    "latitude":iss_la,
    "longitude":iss_long,
    "current_weather":True
}

weather_response = requests.get(weather_url, params=weather_params)
weather_data = weather_response.json()

print("The ISS location temperature is", "\nTemperature:", weather_data["current_weather"]["temperature"])
```
